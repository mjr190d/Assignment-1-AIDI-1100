#### B1. Why Functions Exist [Video]
The same logic is copy-pasted for three students. One of them gets the wrong result.

**Original Code** (run it, don't edit it)

````python
# Ali
ali = [72, 88, 95]
ali_avg = sum(ali) / len(ali)
if ali_avg >= 80:
    print("Ali: Honours", round(ali_avg, 1))
else:
    print("Ali: Pass", round(ali_avg, 1))

# Sam
sam = [64, 70, 81]
sam_avg = sum(sam) / len(sam)
if sam_avg >= 80:
    print("Sam: Honours", round(sam_avg, 1))
else:
    print("Sam: Pass", round(sam_avg, 1))

# Jo
jo = [90, 85, 99]
jo_avg = sum(sam) / len(jo)
if jo_avg >= 80:
    print("Jo: Honours", round(jo_avg, 1))
else:
    print("Jo: Pass", round(jo_avg, 1))
````

**Task:**
- Find and fix the bug.
- Rewrite this as **one function** that is called three times.
- The honours cut-off just changed from `80` to `85`. Update your code.

The problem is with Jo: jo_avg = sum(sam) / len(jo) adds Sam's grades. Jo returns 71.7 and Pass. Her real average is 91.3. This copy paste error is what a function prevents, as the calculation is written once. 

**Your Fix**

````python
HONOURS_CUTOFF = 85

def report(name, grades):
    avg = sum(grades) / len(grades)
    if avg >= HONOURS_CUTOFF:
        status = "Honours"
    else:
        status = "Pass"
    print(f"{name}: {status}", round(avg, 1))

report("Ali", [72, 88, 95])
report("Sam", [64, 70, 81])
report("Jo",  [90, 85, 99])
````

The rewrite has the cut-off updated to 85.
Changes: 1. One function 3 calls. 2. The cut-off is a named constant at the top. 3. Bug can't return since each call passes its own list into grades. So there's no way to mix up Sam's and Jo's data.

#### C1. Why Classes Exist [Video]
Every function needs the `student` passed in again and again.

**Original Code** (run it, don't edit it)

````python
def make_student(name):
    return {"name": name, "grades": []}

def add_grade(student, grade):
    student["grades"].append(grade)

def average(student):
    return sum(student["grades"]) / len(student["grades"])

def report(student):
    print(f"{student['name']}: {average(student):.1f}")


ali = make_student("Ali")
add_grade(ali, 72)
add_grade(ali, 88)
report(ali)
````

**Task:** Rewrite this as a `Student` class so the code at the bottom becomes:

```python
ali = Student("Ali")
ali.add_grade(72)
ali.add_grade(88)
ali.report()
```

Problem: Every function needs the student handed to it. The function is separate from the data, so you have to keep telling it which student to work on. Nothing stops you from passing the wrong thing and we'd only find-out when it crashes or gave the wrong answer. This is the main problem class solves.

**Your Fix**

````python
# C1: your fix here
class Student:
    def __init__(self, name):
        self.name = name
        self.grades = []

    def add_grade(self, grade):
        self.grades.append(grade)

    def average(self):
        return sum(self.grades) / len(self.grades)

    def report(self):
        print(f"{self.name}: {self.average():.1f}")


ali = Student("Ali")
ali.add_grade(72)
ali.add_grade(88)
ali.report()
````

Fix: We still pass the student every time. Python does it through self.

#### C2. Shared vs Per-Object Attributes
Ali and Sam are seeing each other's grades.

**Original Code** (run it, don't edit it)

````python
class Student:
    school = "Humber"
    grades = []

    def __init__(self, name):
        self.name = name

    def add_grade(self, grade):
        self.grades.append(grade)


ali = Student("Ali")
sam = Student("Sam")
ali.add_grade(90)
sam.add_grade(60)
print(ali.name, ali.school, ali.grades)
print(sam.name, sam.school, sam.grades)
````

**Task:** Fix it so each student only sees **their own** grades. `school` should stay **shared** by every student.

**Your Fix**

````python
# C2: your fix here
class Student:
    school = "Humber"          # class attribute: shared by all students

    def __init__(self, name):
        self.name = name       # instance attribute: one per student
        self.grades = []       # instance attribute: a fresh list per student

    def add_grade(self, grade):
        self.grades.append(grade)


ali = Student("Ali")
sam = Student("Sam")
ali.add_grade(90)
sam.add_grade(60)
print(ali.name, ali.school, ali.grades)
print(sam.name, sam.school, sam.grades)
````

#### C3. Methods With and Without `self` [Video]
One call works, the other crashes.

**Original Code** (run it, don't edit it)

````python
class Grader:
    def letter(score):
        if score >= 80:
            return "A"
        elif score >= 65:
            return "B"
        return "C"


print(Grader.letter(85))

g = Grader()
print(g.letter(85))
````

**Task:** Make **both** calls print `A`.

Grader.letter(85) works. g.letter(85) crashes. Called on an instance, ppython automatically passes g as the 1st argument (the self behaviour you've seen). So it actually calls letter(g, 85) but letter only accepts one parameter (TypeError).

**Your Fix**

````python
# C3: your fix here
class Grader:
    @staticmethod
    def letter(score):
        if score >= 80:
            return "A"
        elif score >= 65:
            return "B"
        return "C"


print(Grader.letter(85))

g = Grader()
print(g.letter(85))
````

@staticmethod is a decorator. It fits here because letter doesn't use any data from the object. Adding self (def letter (self, score):) would fix g.letter(85) would crash because 85 would land in self and score would be missing. Only @staticmethod makes both calls work.

#### C4. Edit In Place vs Return
This should print Ali's curved grades. Note: this `Student` takes its grades when it's created.

**Original Code** (run it, don't edit it)

````python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def curve(self, points):
        for i in range(len(self.grades)):
            self.grades[i] += points


ali = Student("Ali", [70, 80, 90])
ali = ali.curve(5)
print(ali.grades)
````

**Task:** Change `curve()` so it **returns a new** `Student` with the curved grades and leaves the original `ali` unchanged. Store the result in a new variable and print both students' grades to prove it.

Expected: `ali.grades` is still `[70, 80, 90]`, and the new student's grades are `[75, 85, 95]`.

**Hint:** Think back to B3. Does creating a new student also create a separate grades list?

This is the grades = grades.sort() trap again that we saw earlier. 

**Your Fix**

````python
# C4: your fix here
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def curve(self, points):
        curved = [g + points for g in self.grades]
        return Student(self.name, curved)


ali = Student("Ali", [70, 80, 90])
ali_curved = ali.curve(5)
print("ali:       ", ali.grades)
print("ali_curved:", ali_curved.grades)
````
